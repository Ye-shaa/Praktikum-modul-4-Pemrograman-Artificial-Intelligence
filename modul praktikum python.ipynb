```python
# No. 5
n = int(input("Masukkan jumlah mahasiswa : "))

print("No.\tNama Mhs\tN.Tugas\tN.Kuis\tN.UTS\tN.UAS\tNilaiAkhir")

for i in range(1, n + 1):
    nama = input(f"Masukkan nama mahasiswa ke-{i} : ")
    tugas = float(input("N.Tugas : "))
    kuis = float(input("N.Kuis : "))
    uts = float(input("N.UTS : "))
    uas = float(input("N.UAS : "))

    nilai_akhir = (tugas + kuis + uts + uas) / 4

    print(f"{i}\t{nama}\t{tugas}\t{kuis}\t{uts}\t{uas}\t{nilai_akhir}")
```

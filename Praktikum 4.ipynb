```python
awal=int(input('Set Nilai Awal = '))
akhir=int(input('Set Nilai Akhir = '))

count=0
summ=0

print('Bilangan antara %d dan %d' %(awal,akhir))

for i in range(awal,akhir+1):
    print(i, end=' ')
    count=count+1
    summ=summ+i

print('Bilangan di atas ada %d bilangan' %count)
print('Jumlah semua bilangan adalah %d' %summ)
```

```python
banyakMhs = int(input('Banyak Mahasiswa : '))
for M in range (banyakMhs):
    print('Mahasiswa ke-',M+1)
    banyakMK = int(input('Banyak Matakuliah yang diambil : '))
    totalnilai=0
    for N in range(banyakMK):
        nilai=int(input('Input Nilai ke %d : ' %(N+1)))
        totalnilai=totalnilai+nilai
    rerata=totalnilai/banyakMK
    print('Rata-Rata : ',rerata)
```

```python
# LATIHAN 1 - Menghitung total pembelian N buah barang
N = int(input("Masukkan jumlah jenis barang (N): "))
total_belanja = 0

for i in range(1, N+1):
    print(f"\nBarang ke-{i}")
    harga = int(input(f"  Harga barang ke-{i}: Rp "))
    jumlah = int(input(f"  Jumlah barang ke-{i}: "))
    subtotal = harga * jumlah
    print(f"  Subtotal: Rp {subtotal}")
    total_belanja = total_belanja + subtotal  # atau total_belanja += subtotal

print(f"\nTotal belanja {N} jenis barang adalah: Rp {total_belanja}")
```

```python
#LATIHAN 2
N = int(input("Masukkan banyaknya bilangan (N): "))

genap = 0
ganjil = 0
daftar_genap = []
daftar_ganjil = []


for i in range(N):
    angka = int(input(f"Masukkan bilangan ke-{i+1}: "))
    if angka % 2 == 0:
        genap += 1
        daftar_genap.append(angka)
    else:
        ganjil += 1
        daftar_ganjil.append(angka)

print("\n=== HASIL ===")
print(f"Bilangan genap  : {daftar_genap}")
print(f"Jumlah genap    : {genap}")
print(f"Bilangan ganjil : {daftar_ganjil}")
print(f"Jumlah ganjil   : {ganjil}")
```

```python
# Program menampilkan bilangan kelipatan 3

awal = int(input("Masukkan bilangan awal: "))
akhir = int(input("Masukkan bilangan akhir: "))

print("Bilangan kelipatan 3:")

for i in range(awal, akhir + 1):
    if i % 3 == 0:
        print(i, end=" ")
```
